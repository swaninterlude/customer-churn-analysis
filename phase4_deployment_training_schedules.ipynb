# Ulta Curbside – Phase 4 Implementation Schedules

This notebook generates Gantt charts for:
- **Deployment Schedule**
- **Training Schedule**

The dates and phases are based on your Unit 3 Design Phase document (pilot, training & optimization, regional rollout, review/feedback) and are meant for **Phase 4 – Implementation** in Unit 4.

You can change task names, dates, and colors to match any updates to your report.

```python
# 2. Imports and color settings
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.dates as mdates
from datetime import datetime

# You can change these colors to anything you like (or set them to None for default)
DEPLOY_COLOR = '#5DADE2'   # Deployment bars color
TRAIN_COLOR = '#58D68D'    # Training bars color

def make_gantt(df, title, filename, bar_color=None):
    fig, ax = plt.subplots(figsize=(10, 4))
    for _, row in df.iterrows():
        duration = (row['End'] - row['Start']).days
        ax.barh(row['Task'], duration, left=row['Start'], color=bar_color)

    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=1))
    plt.xticks(rotation=45)
    plt.title(title)
    plt.tight_layout()

    # Show chart in Colab
    plt.show()

    # Save chart as a PNG so you can download and insert into Word
    fig.savefig(filename, dpi=300, bbox_inches='tight')
    plt.close(fig)

    print(f'Saved chart as: {filename}')
```
