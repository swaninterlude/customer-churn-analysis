```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

1. Correlation Heatmap: The heatmap reveals a moderate correlation (-0.39) between satisfaction_level and left. This suggests that employees with lower satisfaction levels are more likely to leave the company. Conversely, time_spend_company shows a moderate positive correlation (0.14) with left, including that employees who have spent more years at the company might be more prone to leaving.

```python
df_corr = df[['satisfaction_level', 'last_evaluation', 'number_project', 'average_montly_hours','time_spend_company','Work_accident','left','promotion_last_5years']]
```

```python
plt.figure(figsize=(14,12))
sns.heatmap(df_corr.corr(), annot=True, cmap='Blues')
```

Distribution Plot of Satisfaction Level: The distribution plot of satisfaction_level reveals a distinct bimodal shape. There are two prominent peaks: one at a lower satisfaction level (around 0.1 to 0.2) and another at a higher satisfaction level (around 0.7 to 0.9). This suggests that employees tend to fall into two main groups: those who are quite dissatisfied and those who are quite satisfied. The relatively low number of employees with mid-range satisfaction levels could be an interesting point for further investigation into the factors driving extreme satisfaction or dissatisfaction.

```python
sns.histplot(df['satisfaction_level'], kde=True,bins=30, color='red')
plt.title('Distribution Plot of Satisfaction Level')
plt.show()
```

Distribution Plot of Last Evaluation: The distribution plot for last_evaluation also shows a bimodal distribution, with notabe peals around 0.5 and 0.8. This suggests that employee evaluations tend to cluster at two different performance levels. One group of employees receives evaluations around the midpoint, while another group receives higher evaluations. Understanding the characteristics of employees in these two groups could provide insights into performance management withing the company.

```python
sns.histplot(df['last_evaluation'], kde=True,bins=30, color='red')
plt.title('Distribution Plot of Last Evaluation')
plt.show()
```

Distribution Plot of Employee Average Monthly Hours: The distribution plot of average_montly_hours appears to be roughly normally distributed, with a peak around 150 hours.
```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df['average_montly_hours'], kde=True,bins=30, color="red")
plt.title('Distribution of Employee Average Monthly Hours')
plt.show()
```

Outlier

```python
olm=[]
for c in df.columns[0:7]:
  q1= df[c].quantile(0.25)
  q3 = df[c].quantile(0.75)
  iqr = q3 - q1 #Interquartile range
  fence_low = q3-1.5*iqr
  fence_high = q3+1.5*iqr
  lower_out = len(df.loc[(df[c] < fence_low) , c])
  upper_out = len(df.loc[(df[c] > fence_high) , c])
  outlier_count = upper_out+lower_out
  prop_out = outlier_count/len(df)
  olm.append([c, prop_out])
  print(f"{c}: {prop_out:.2%}")
```
