# Project Proposal: Predictive Analysis of Employee Retention Using Machine Learning Techniques

## 1. Problem Statement
Employee turnover is a costly and persistent issue for many organizations, often cause by factors that are not visible. Traditional HR methods struggle to predict which employees are likely to leave. This project aims to use machine learning and data analytics to identify key drivers of employee retention and improve prediction accuracy.

## 2. Objective and Success Metrics
Objective: Develop and compare machine learning models to predict employee retention with at least 85% accuracy, completed within the current term.

Success Metric: The project will be considered successful if the best performing model correctly predicts employee retention with at least 85% accuracy.

## 3. System Inputs and Outputs
Data Source: The dataset used in this project is a structures HR employee dataset containing information about employees who have either left or stayed at a company. It includes features such as satisfaction level, number of projects, average monthly hours, time spent at the company, promotion in the last five yeaers, department, and salary level.

Data Source: CSV

Primary Output: The primary output of the project will be an evaluated machine learning model capable of predicting employee retention.

Output Format: The results will be presented in the form of performance metrics, along with visualizations such as bar charts and model comparison tables.

## 4. Scope

**In-Scope Features:**
- Injest and clean HR dataset
- Perform exploratory data analysis
- Build and compare machine learning models

**Out-of-Scope Features:**
- Predicting employee performance or activity
- Including external factors like economic conditions
- Building real-time retention prediction analysis

## 5. High Level Plan and Milestones
1. Data Collection and Cleaning: Import HR dataset and encode categorical variables.
2. Exploratory Data Analysis: Analyze data to identify patterns and key factors affecting employee retention.
3. Model Development: Implement and train support vector machine and multilayer perceptron models using the processed data.
4. Model Evauation and Comparison: Evaluate models using accuracy, precision, recall, and F1 score; compare results to identify the best performer.
5. Final Report and Recommendations: Document findings, visualizations, model comparisons, and provide actionable insights for HR decision-making.
